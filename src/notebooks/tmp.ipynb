```python
import json
import pandas as pd

# CONFIG
INPUT_JSON = "C:/Users/Rosario/Desktop/dev/Nursind/failed/text_extraction/broken.json"     # file JSON locale
OUTPUT_CSV = "output.csv"     # file CSV di output
ID_FIELD = "file_id"
LINK_FIELD = "drive_link"

# Load JSON
with open(INPUT_JSON, "r", encoding="utf-8") as f:
    data = json.load(f)

# To DataFrame
df = pd.DataFrame(data)

# Build Drive link
df[LINK_FIELD] = df[ID_FIELD].apply(
    lambda x: f"https://drive.google.com/file/d/{x}/view" if pd.notna(x) else ""
)

# Save CSV
df.to_csv(OUTPUT_CSV, index=False, encoding="utf-8")

print(f"✅ CSV creato: {OUTPUT_CSV}")
```

```python
def fill_item_report_item(file, included_index):
    _,_, employee, filename_txt = file["source_txt"].split("\\")
    filename = filename_txt.replace(".txt", ".pdf")
    index_file = next( file for file in included_index.files if file.file_name == filename)
    if not index_file:
        print(f"File {filename} not found in included index")
        return file
    new_item = { "employee": index_file.employee,"employee_id": index_file.employee_id, "index_file_name": index_file.file_name, "index_file_id": index_file.file_id } | file
    return new_item


new_failed_parsing_report = [ fill_item_report_item(file, included_index) for file  in failed_parsing_report ]

with open("../../output/parsed_from_text/files_without_days_fixed.json", "w", encoding="utf-8") as f:
    json.dump(new_failed_parsing_report, f, indent=4, ensure_ascii=False)
```
