# Extract Events From Text

This notebook reads the shared text extraction output and generates per-file `events_from_text_raw.csv` artifacts.


```python
from pathlib import Path
import sys

repo_root = Path.cwd().resolve()
while not (repo_root / "src").exists() and repo_root != repo_root.parent:
    repo_root = repo_root.parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

from src.notebooks.shared_config import load_notebook_context

ctx = load_notebook_context()
paths = ctx.paths
events_cfg = ctx.step("extract_events")

{
    "config_path": str(ctx.config_path),
    "input_dir": str(paths.documents_output),
    "output_dir": str(paths.events_output),
}
```

```python
from src.drive_service.logging_utils import setup_logging
from src.extract_events_from_text_raw.options import ExtractEventsFromTextOptions

VERBOSE = True
report_path = paths.events_output / events_cfg["report_name"]

options = ExtractEventsFromTextOptions(
    input_dir=str(paths.documents_output),
    output_dir=str(paths.events_output),
    out_name=str(events_cfg["out_name"]),
    report_json=str(report_path),
    verbose=VERBOSE,
)
setup_logging(VERBOSE)
options
```

```python
from src.extract_events_from_text_raw.runtime import run_from_options

report = run_from_options(options)

{
    "report_json": str(report_path),
    "files_ok": report.get("files_ok"),
    "files_failed": report.get("files_failed"),
}
```
